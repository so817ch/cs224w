```python
import networkx as nx

# 左图 G
G = nx.Graph()
G.add_edges_from([
    ('A', 'B'), ('A', 'C'), ('B', 'D'),
    ('C', 'E'), ('D', 'F'), ('E', 'F'),
    ('E', 'G'), ('G', 'I'), ('I', 'J'),
    ('J', 'H'), ('H', 'F'),
])

# 右图 G_hat
G_hat = nx.Graph()
G_hat.add_edges_from([
    ('A', 'B'), ('A', 'C'), ('B', 'D'),
    ('C', 'E'), ('D', 'E'), ('E', 'F'),
    ('F', 'G'), ('G', 'I'), ('I', 'J'),
    ('J', 'H'), ('H', 'F'),
])

print('G:', G)
print('G_hat:', G_hat)
```

```python
g=nx.to_numpy_array(G)
g_hat=nx.to_numpy_array(G_hat)
g_eigenvalues, g_V=np.linalg.eigh(g)
g_hat_eigenvalues, g_hat_V=np.linalg.eigh(g_hat)
g_eigenvalues, g_hat_eigenvalues,g_V,g_hat_V
```
